# Limpieza de la base de datos de Interlub

## Cargar la base de datos

```python
import pandas as pd

# Cargas los datos
df = pd.read_excel('../data/raw/DATOS_TEC.xlsx')
df.head()
```

## Limpieza básica

```python
# Crear una copia
data = df.copy()
data.head()
```

```python
# Eliminar la columna de "Unnamed: 0" creada por Excel
data.drop(columns = ['Unnamed: 0'], inplace = True)
data.head()
```

```python
# Eliminar columnas "no relevantes para la recomendación"
relevant_columns = ['CodigoProducto', 'Familia', 'Aceite_Base', 'Espesante',
                    'Grado_ISO_VG', 'Viscosidad_a_40°C_cSt', 'Viscosidad_a_100°C_cSt',
                    'Indice_de_Viscosidad', 'Punto_de_Flama_°C', 'Punto_de_Congelamiento_°C',
                    'Peso_Específico', 'Grado_NLGI_Consistencia',
                    'Penetración_de_Cono_a_25°C_0.1mm', 'Estabilidad_Mecánica_%',
                    'Viscosidad_del_Aceite_Base_a_40°C._cSt', 'Punto_de_Gota_°C',
                    'Resistencia_al_Lavado_por_Agua_a_80°C_%', 'Factor_de_Velocidad',
                    'Desgaste_Cuatro_Bolas_mm', 'Punto_de_Soldadura_Cuatro_Bolas_kgf',
                    'Carga_Timken_Ok_lb', 'Corrosión_al_Cobre', 'Registro_NSF',
                    'Temperatura_de_Servicio_°C_min', 'Temperatura_de_Servicio_°C_max',
                    'Color']

data = data[relevant_columns]
data.head()
```

```python
# Verificar que todos los productos son únicos
productos_unicos = len(data['CodigoProducto'].unique())
print(f'Productos únicos: {productos_unicos}\nFilas en el dataset: {data.shape[0]}')
```

```python
# Verificar duplicados
print(f'¿Existen duplicados?: {data.duplicated().any()}')
```

```python
# Verificar tipos de datos
print(data.info())
```

```python
# Identificar la manera en que se representan los datos para ver por qué son strings
str_columns = data.columns[data.dtypes == 'str']
for col in str_columns[4:len(str_columns) - 1]:
    s = data[col]
    print(f"\n== {repr(col)}")
    print(s.str.replace(r"\d", "9", regex=True).value_counts().to_dict())
```
