## Cargar la base de datos

```python
%load_ext autoreload
%autoreload 2

from reto_multivariado.limpieza import limpiar, COLUMNAS_RELEVANTES, COLS_NO_NUMERICAS, RENOMBRES

import pandas as pd
```

```python
# Importar la base de datos en crudo
df = pd.read_excel("../data/raw/DATOS_TEC.xlsx")
df.head()
```

### Exploración de la base cruda

```python
data = df[COLUMNAS_RELEVANTES]
data.head()
```

```python
no_numericas_originales = [k for k, v in RENOMBRES.items() if v in COLS_NO_NUMERICAS]
str_columns = [c for c in data.columns[data.dtypes == "str"] if c not in no_numericas_originales]
for col in str_columns:
    s = data[col]
    print(f"\n== {col!r}")
    print(s.str.replace(r"\d", "9", regex=True).value_counts().to_dict())
```

### Limpieza

```python
data_limpia = limpiar(df)
```
